### 2. Setting up Roles and User accounts

- You need to have certain roles in a db first
- The you need users with certain permissions
- Then you can connect with certain previlages using certain credentials

1. Admin Roles
    - Users with this  role may do whatever they wish
    - we will create a `dbadmin_role`
        - 1 Admin user account: `idrees_admin`

2. Learner Account:
    - Users with this role may just interact with data
    - we will create a `learner_role`
        - 1 Learner user account: `idrees_learner`

```python
-- Create the administrator role
CREATE ROLE 'dbadmin_role';

-- Administrator: all privileges on the ecommerce database
GRANT ALL PRIVILEGES ON shopoby_db.* TO 'dbadmin_role';

-- Create your administrator account
CREATE USER 'idrees_admin'@'%'
IDENTIFIED BY 'ReplaceWithAStrongAdminPassword!';

-- Assign roles
GRANT 'dbadmin_role' TO 'idrees_admin'@'%';

-- Make each role the default role
SET DEFAULT ROLE 'idrees_admin' TO 'idrees'@'%';
```



The '%' in user name suggests that:
- Connections from any matching host can access this user
- alternatives could have been:

Account |	Meaning
--|--
'idrees'@'localhost'	 | Connections from the database server itself
'idrees'@'192.168.1.10'	| Connections from the specified host
'idrees'@'192.168.1.%' | 	Connections from matching hosts in that IP range
'idrees'@'%'	| Connections from any matching host

```python
-- Create a role for students
CREATE ROLE 'learner_role';

-- Students: read and modify records
GRANT SELECT, INSERT, UPDATE, DELETE
ON shopoby_db.* TO 'learner_role';

-- Create a student account
CREATE USER '2hg26csUn8Q3kYy.idrees_learner'@'%'
IDENTIFIED BY 'idrees_learner123';

-- Assign roles
GRANT 'learner_role' TO '2hg26csUn8Q3kYy.idrees_learner'@'%';

-- Make each role the default role
SET DEFAULT ROLE 'learner_role' TO '2hg26csUn8Q3kYy.idrees_learner'@'%';
```

```python
-- Check the permissions for users and roles

SHOW GRANTS FOR '2hg26csUn8Q3kYy.idrees_admin';
SHOW GRANTS FOR '2hg26csUn8Q3kYy.idrees_learner';

SHOW GRANTS FOR 'dbadmin_role';
SHOW GRANTS FOR 'learner_role';
```
